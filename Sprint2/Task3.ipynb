```
using System;

public interface IShippable
{
    double Weight { get; set; }
    double CalculateShipping();
}

public interface IWarrantable
{
    int WarrantyPeriod { get; set; }
    void ExtendWarranty(int months);
}

public interface IReturnable
{
    int ReturnDays { get; set; }
    bool CanReturn(int daysSincePurchase);
}

public class Product
{
    private string _name;
    private double _price;
    private string _manufacturer;
    private int _quantity;
    private string _category;
    private string _article;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public double Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не может быть отрицательной!");
        }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = value; }
    }

    public int Quantity
    {
        get { return _quantity; }
        set
        {
            if (value >= 0)
                _quantity = value;
            else
                throw new ArgumentOutOfRangeException("Количество не может быть отрицательным!");
        }
    }

    public string Category
    {
        get { return _category; }
        set { _category = value; }
    }

    public string Article
    {
        get { return _article; }
        set { _article = value; }
    }

    public Product(string name, double price, string manufacturer, int quantity, string category, string article)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Quantity = quantity;
        Category = category;
        Article = article;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, Категория: {Category}, Артикул: {Article}, Остаток: {Quantity} шт.";
    }

    public virtual void Discount(double percent)
    {
        Price -= Price * percent / 100;
        Console.WriteLine($"Применена скидка {percent}%. Новая цена: {Price} руб.");
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public void AddStock(int amount)
    {
        Quantity += amount;
        Console.WriteLine($"Поступление товара «{Name}»: +{amount} шт. Остаток: {Quantity} шт.");
    }

    public bool Sell(int amount)
    {
        if (amount > Quantity)
        {
            Console.WriteLine($"Недостаточно товара «{Name}» на складе. Доступно: {Quantity} шт.");
            return false;
        }
        Quantity -= amount;
        Console.WriteLine($"Продано {amount} шт. товара «{Name}». Остаток: {Quantity} шт.");
        return true;
    }

    public double GetTotalValue()
    {
        return Price * Quantity;
    }

    public void ComparePriceWith(Product otherProduct)
    {
        if (Price > otherProduct.Price)
        {
            Console.WriteLine($"Товар «{Name}» дороже, чем «{otherProduct.Name}» на {Price - otherProduct.Price} руб.");
        }
        else if (Price < otherProduct.Price)
        {
            Console.WriteLine($"Товар «{Name}» дешевле, чем «{otherProduct.Name}» на {otherProduct.Price - Price} руб.");
        }
        else
        {
            Console.WriteLine($"Товары «{Name}» и «{otherProduct.Name}» стоят одинаково.");
        }
    }

    public void ApplyBulkDiscount(Product secondProduct, double bundleDiscountPercent)
    {
        Console.WriteLine($"\n[Акция: Покупка комплекта «{Name}» + «{secondProduct.Name}»]");
        Discount(bundleDiscountPercent);
        secondProduct.Discount(bundleDiscountPercent);
    }
}

public class Electronics : Product, IShippable, IWarrantable
{
    private int _warrantyPeriod;
    private double _weight;
    private int _powerConsumption;
    private string _model;

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set
        {
            if (value >= 0)
                _warrantyPeriod = value;
            else
                throw new ArgumentOutOfRangeException("Гарантия не может быть отрицательной!");
        }
    }

    public double Weight
    {
        get { return _weight; }
        set { _weight = value; }
    }

    public int PowerConsumption
    {
        get { return _powerConsumption; }
        set { _powerConsumption = value; }
    }

    public string Model
    {
        get { return _model; }
        set { _model = value; }
    }

    public Electronics(string name, double price, string manufacturer, int quantity, string article,
        int warrantyPeriod, double weight, int powerConsumption, string model)
        : base(name, price, manufacturer, quantity, "Электроника", article)
    {
        WarrantyPeriod = warrantyPeriod;
        Weight = weight;
        PowerConsumption = powerConsumption;
        Model = model;
    }

    public override void Discount(double percent)
    {
        double extra = WarrantyPeriod > 12 ? 5 : 0;
        base.Discount(percent + extra);
        Console.WriteLine($"(Учтена гарантия {WarrantyPeriod} мес.)");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Модель: {Model}, Гарантия: {WarrantyPeriod} мес., Вес: {Weight} кг, Мощность: {PowerConsumption} Вт";
    }

    public void ExtendWarranty(int months)
    {
        WarrantyPeriod += months;
        Console.WriteLine($"Гарантия на «{Name}» продлена на {months} мес. Всего: {WarrantyPeriod} мес.");
    }

    public double CalculateShipping()
    {
        return 300 + Weight * 50;
    }

    public double GetYearlyEnergyCost(double tariff, double hoursPerDay)
    {
        return PowerConsumption / 1000.0 * hoursPerDay * 365 * tariff;
    }
}

public class Smartphone : Electronics
{
    private int _memory;
    private int _batteryCapacity;
    private string _operatingSystem;
    private double _screenSize;

    public int Memory
    {
        get { return _memory; }
        set { _memory = value; }
    }

    public int BatteryCapacity
    {
        get { return _batteryCapacity; }
        set { _batteryCapacity = value; }
    }

    public string OperatingSystem
    {
        get { return _operatingSystem; }
        set { _operatingSystem = value; }
    }

    public double ScreenSize
    {
        get { return _screenSize; }
        set { _screenSize = value; }
    }

    public Smartphone(string name, double price, string manufacturer, int quantity, string article,
        int warrantyPeriod, double weight, int powerConsumption, string model,
        int memory, int batteryCapacity, string operatingSystem, double screenSize)
        : base(name, price, manufacturer, quantity, article, warrantyPeriod, weight, powerConsumption, model)
    {
        Memory = memory;
        BatteryCapacity = batteryCapacity;
        OperatingSystem = operatingSystem;
        ScreenSize = screenSize;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Память: {Memory} ГБ, Батарея: {BatteryCapacity} мАч, ОС: {OperatingSystem}, Экран: {ScreenSize}\"";
    }

    public double GetTalkTime()
    {
        return BatteryCapacity / 250.0;
    }

    public void InstallApp(string appName, int sizeMb)
    {
        double sizeGb = sizeMb / 1024.0;
        if (sizeGb > Memory)
        {
            Console.WriteLine($"Недостаточно памяти для установки «{appName}».");
            return;
        }
        Console.WriteLine($"Приложение «{appName}» установлено на «{Name}» ({OperatingSystem}).");
    }
}

public class Clothing : Product, IShippable, IReturnable
{
    private string _size;
    private string _material;
    private string _color;
    private string _season;
    private double _weight;
    private int _returnDays;

    public string Size
    {
        get { return _size; }
        set { _size = value; }
    }

    public string Material
    {
        get { return _material; }
        set { _material = value; }
    }

    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    public string Season
    {
        get { return _season; }
        set { _season = value; }
    }

    public double Weight
    {
        get { return _weight; }
        set { _weight = value; }
    }

    public int ReturnDays
    {
        get { return _returnDays; }
        set { _returnDays = value; }
    }

    public Clothing(string name, double price, string manufacturer, int quantity, string article,
        string size, string material, string color, string season, double weight)
        : base(name, price, manufacturer, quantity, "Одежда", article)
    {
        Size = size;
        Material = material;
        Color = color;
        Season = season;
        Weight = weight;
        ReturnDays = 14;
    }

    public override void Display()
    {
        Console.WriteLine(GetInfo() + $", Размер: {Size}, Материал: {Material}, Цвет: {Color}, Сезон: {Season}");
    }

    public double CalculateShipping()
    {
        return 150 + Weight * 30;
    }

    public bool CanReturn(int daysSincePurchase)
    {
        return daysSincePurchase <= ReturnDays;
    }

    public bool IsSuitableForSeason(string season)
    {
        return Season == season;
    }
}

public class Books : Product, IReturnable
{
    private string _author;
    private int _pages;
    private string _genre;
    private int _year;
    private int _returnDays;

    public string Author
    {
        get { return _author; }
        set { _author = value; }
    }

    public int Pages
    {
        get { return _pages; }
        set { _pages = value; }
    }

    public string Genre
    {
        get { return _genre; }
        set { _genre = value; }
    }

    public int Year
    {
        get { return _year; }
        set { _year = value; }
    }

    public int ReturnDays
    {
        get { return _returnDays; }
        set { _returnDays = value; }
    }

    public Books(string name, double price, string manufacturer, int quantity, string article,
        string author, int pages, string genre, int year)
        : base(name, price, manufacturer, quantity, "Книги", article)
    {
        Author = author;
        Pages = pages;
        Genre = genre;
        Year = year;
        ReturnDays = 7;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}, Страниц: {Pages}, Жанр: {Genre}, Год: {Year}";
    }

    public bool CanReturn(int daysSincePurchase)
    {
        return daysSincePurchase <= ReturnDays;
    }

    public double GetReadingTime(int pagesPerHour)
    {
        return (double)Pages / pagesPerHour;
    }
}

public class EBook : Books, IWarrantable
{
    private double _fileSize;
    private string _format;
    private int _warrantyPeriod;

    public double FileSize
    {
        get { return _fileSize; }
        set { _fileSize = value; }
    }

    public string Format
    {
        get { return _format; }
        set { _format = value; }
    }

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set { _warrantyPeriod = value; }
    }

    public EBook(string name, double price, string manufacturer, int quantity, string article,
        string author, int pages, string genre, int year, double fileSize, string format)
        : base(name, price, manufacturer, quantity, article, author, pages, genre, year)
    {
        FileSize = fileSize;
        Format = format;
        WarrantyPeriod = 6;
        ReturnDays = 0;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Формат: {Format}, Размер файла: {FileSize} МБ";
    }

    public void ExtendWarranty(int months)
    {
        WarrantyPeriod += months;
        Console.WriteLine($"Доступ к загрузке «{Name}» продлён на {months} мес. Всего: {WarrantyPeriod} мес.");
    }

    public double GetDownloadTime(double speedMbps)
    {
        return FileSize * 8 / speedMbps;
    }
}

Smartphone phone = new Smartphone("Смартфон", 50000, "Samsung", 20, "EL-001", 24, 0.2, 15, "Galaxy S", 256, 5000, "Android", 6.7);
phone.Display();
phone.Discount(10);
phone.ExtendWarranty(6);
Console.WriteLine($"Доставка: {phone.CalculateShipping()} руб.");
Console.WriteLine($"Время разговора: {phone.GetTalkTime()} ч.");
phone.InstallApp("Telegram", 200);

Console.WriteLine();

Clothing tshirt = new Clothing("Футболка", 1500, "Nike", 50, "CL-001", "L", "Хлопок", "Чёрный", "Лето", 0.3);
tshirt.Display();
tshirt.Discount(15);
Console.WriteLine($"Доставка: {tshirt.CalculateShipping()} руб.");
Console.WriteLine($"Возврат через 10 дней: {tshirt.CanReturn(10)}");
Console.WriteLine($"Подходит для лета: {tshirt.IsSuitableForSeason("Лето")}");

Console.WriteLine();

Books book = new Books("Война и мир", 800, "Эксмо", 30, "BK-001", "Л.Н. Толстой", 1300, "Роман", 1869);
book.Display();
book.Discount(5);
Console.WriteLine($"Возврат через 10 дней: {book.CanReturn(10)}");
Console.WriteLine($"Время чтения: {book.GetReadingTime(30)} ч.");

Console.WriteLine();

EBook ebook = new EBook("Преступление и наказание", 400, "Эксмо", 100, "BK-002", "Ф.М. Достоевский", 600, "Роман", 1866, 3.5, "EPUB");
ebook.Display();
ebook.ExtendWarranty(3);
Console.WriteLine($"Время загрузки: {ebook.GetDownloadTime(20)} сек.");

Console.WriteLine("\n===== Демонстрация взаимодействия объектов =====");
phone.ComparePriceWith(tshirt);
tshirt.ComparePriceWith(book);
tshirt.ApplyBulkDiscount(book, 10);

Console.WriteLine();
phone.Sell(5);
phone.AddStock(10);
Console.WriteLine($"Стоимость остатков: {phone.GetTotalValue()} руб.");
```
