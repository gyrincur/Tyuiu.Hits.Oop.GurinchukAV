```
using System;

public class Product
{
    private string _name;
    private double _price;
    private string _manufacturer;

    // Свойства с полноценными геттерами и сеттерами
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public double Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не может быть отрицательной!");
        }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = value; }
    }

    // Конструктор базового класса
    public Product(string name, double price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}";
    }

    public virtual void Discount(double percent)
    {
        Price -= Price * percent / 100;
        Console.WriteLine($"Применена скидка {percent}%. Новая цена: {Price} руб.");
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    // Взаимодействие объектов: сравнение стоимости товаров
    public void ComparePriceWith(Product otherProduct)
    {
        if (Price > otherProduct.Price)
        {
            Console.WriteLine($"Товар «{Name}» дороже, чем «{otherProduct.Name}» на {Price - otherProduct.Price} руб.");
        }
        else if (Price < otherProduct.Price)
        {
            Console.WriteLine($"Товар «{Name}» дешевле, чем «{otherProduct.Name}» на {otherProduct.Price - Price} руб.");
        }
        else
        {
            Console.WriteLine($"Товары «{Name}» и «{otherProduct.Name}» стоят одинаково.");
        }
    }

    // Взаимодействие объектов: совместная скидка при покупке комплекта из двух товаров
    public void ApplyBulkDiscount(Product secondProduct, double bundleDiscountPercent)
    {
        Console.WriteLine($"\n[Акция: Покупка комплекта «{Name}» + «{secondProduct.Name}»]");
        this.Discount(bundleDiscountPercent);
        secondProduct.Discount(bundleDiscountPercent);
    }
}

// 1. Электроника
public class Electronics : Product
{
    private int _warrantyPeriod;

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set
        {
            if (value >= 0)
                _warrantyPeriod = value;
            else
                throw new ArgumentOutOfRangeException("Гарантия не может быть отрицательной!");
        }
    }

    public Electronics(string name, double price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
    }

    public override void Discount(double percent)
    {
        double extra = WarrantyPeriod > 12 ? 5 : 0;
        base.Discount(percent + extra);
        Console.WriteLine($"(Учтена гарантия {WarrantyPeriod} мес.)");
    }
}

// 2. Одежда
public class Clothing : Product
{
    private string _size;

    public string Size
    {
        get { return _size; }
        set { _size = value; }
    }

    public Clothing(string name, double price, string manufacturer, string size)
        : base(name, price, manufacturer)
    {
        Size = size;
    }

    public override void Display()
    {
        Console.WriteLine(GetInfo() + $", Размер: {Size}");
    }
}

// 3. Книги
public class Books : Product
{
    private string _author;

    public string Author
    {
        get { return _author; }
        set { _author = value; }
    }

    public Books(string name, double price, string manufacturer, string author)
        : base(name, price, manufacturer)
    {
        Author = author;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}";
    }
}

// ===== Запуск =====

Electronics phone = new Electronics("Смартфон", 50000, "Samsung", 24);
phone.Display();
phone.Discount(10);

Console.WriteLine();

Clothing tshirt = new Clothing("Футболка", 1500, "Nike", "L");
tshirt.Display();
tshirt.Discount(15);

Console.WriteLine();

Books book = new Books("Война и мир", 800, "Эксмо", "Л.Н. Толстой");
book.Display();
book.Discount(5);

Console.WriteLine("\n===== Демонстрация взаимодействия объектов =====");
phone.ComparePriceWith(tshirt);
tshirt.ComparePriceWith(book);

// Взаимодействие при совместной покупке
tshirt.ApplyBulkDiscount(book, 10);
```
